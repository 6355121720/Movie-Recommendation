```python
import os
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
os.listdir('1950-2019')
```

```python
data1 = pd.read_csv('1950-2019/bollywood_full_1950-2019.csv')
data1.head()
```

```python
data2 = pd.read_csv('1950-2019/bollywood_crew_data_1950-2019.csv')
data2.head()
```

```python
data3 = pd.read_csv('1950-2019/bollywood_crew_1950-2019.csv')
data3.head()
```

```python
data4 = pd.read_csv('1950-2019/bollywood_writers_data_1950-2019.csv')
data4.head()
```

```python
data = data1.loc[:,['title_y','imdb_id','is_adult','year_of_release','genres','summary','actors']]
data.head()
```

```python
data = data.merge(data3,on='imdb_id',how='left')
```

```python
data.head()
```

```python
data.replace('\\N',np.nan,inplace=True)
```

```python
data.isnull().sum()
```

```python
data.dropna(inplace=True)
```

```python
data.isnull().sum()
```

```python
data.duplicated().sum()
```

```python
data.drop_duplicates(inplace=True)
```

```python
data.duplicated().sum()
```

```python
data.duplicated(subset=['title_y']).sum()
```

```python
data.drop_duplicates(subset=['title_y'],inplace=True)
```

```python
data['is_adult'].unique()
```

```python
data.drop(columns=['is_adult'],inplace=True)
```

```python
data.head(3)
```

```python
def classify_year(year):
  return year[2]*2
```

```python
data['year_of_release'].unique()
```

```python
data['year_of_release'] = data['year_of_release'].apply(classify_year)
```

```python
data['year_of_release'].unique()
```

```python
data['genres'] = data['genres'].apply(lambda x : x.split('|'))
```

```python
data['summary'] = data['summary'].apply(lambda x : x.split(' '))
```

```python
data['actors'] = data['actors'].apply(lambda x : x.split('|'))
```

```python
data['directors'] = data['directors'].apply(lambda x : x.split('|'))
```

```python
data['writers'] = data['writers'].apply(lambda x : x.split('|'))
```

```python
data['year_of_release'] = data['year_of_release'].apply(lambda x : [x])
```

```python
data.head()
```

```python
data.set_index(np.array([i for i in range(data.shape[0])]),inplace=True)
```

```python
data['tags'] = (data['year_of_release']+data['genres']+data['summary']+data['actors']+data['directors']+data['writers'])
```

```python
data.head()
```

```python
data = data.loc[:,['title_y','imdb_id','tags']]
```

```python
data['tags'] = data['tags'].apply(lambda x : [str(i).lower() for i in x if i!=''])
```

```python
data['tags'] = data['tags'].apply(lambda x : " ".join(x))
```

```python
data.head()
```

```python
from nltk.stem.porter import PorterStemmer
```

```python
ps = PorterStemmer()
```

```python
def stem(s):
  l=[]
  for i in s.split(' '):
    l.append(ps.stem(i))
  return " ".join(l)
```

```python
data['tags'] = data['tags'].apply(stem)
```

```python
data.head()
```

```python
from sklearn.feature_extraction.text import CountVectorizer
```

```python
cv = CountVectorizer(max_features=5000,stop_words='english')
```

```python
movie_tags = cv.fit_transform(data['tags']).toarray()
movie_tags
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
movie_similarity = cosine_similarity(movie_tags)
```

```python
movie_similarity.shape
```

```python
def get_recommendation(movie_name):
  movies=sorted(list(enumerate(movie_similarity[data[data['title_y']==movie_name].index[0]]))
    ,reverse=True,key=lambda x : x[1])
  indices=[]
  for i in range(1,6):
    indices.append(list(data.loc[movies[i][0],['title_y','imdb_id']]))
  return indices
```

```python
get_recommendation('PK')
```
